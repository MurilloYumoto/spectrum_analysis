```python
def normalize_sample(spectrum_sample):
    """Realiza a normalização Min-Max do espectro.
    Args:
        spectrum_sample (pd.Series): Espectro bruto da amostra.
    Returns:
        pd.Series: Espectro normalizado com valores entre 0 e 1.
    """
    # spectrum_sample = spectrum_sample.astype(float)
    result  = []
    for value in spectrum_sample:
        norm_spectrum = (value - min(spectrum_sample)) / (max(spectrum_sample) - min(spectrum_sample))
        result.append(norm_spectrum)
    return result



df['spectrum'] = df['spectrum'].apply(normalize_sample)
```

### Beverage_type / Specie_type specs

```python
def plot_curve_mean(df,
                    column,
                    description: str):
    

    mean_spec1, std_spec = {}, {}

    for type, specs in df.groupby(column)['spectrum']:
        stack = np.vstack(specs.values)
        mean_spec1[type] = stack.mean(axis=0)
        # Desvio padrão amostral
        std_spec[type] = stack.std(axis=0, ddof=1)
        
    n_bands = np.arange(288)

    fig = go.Figure()
    for type in mean_spec1:
        y_mean = mean_spec1[type]
        y_std = std_spec[type]
        color = COLORS[column].get(type, '#999999')
        
        def _hex_to_rgb(hex_color):
            hex_color = hex_color.lstrip("#")
            return ("rgba" + str(tuple(int(hex_color[i:i+2], 16) for i in (0, 2, 4)))).replace(")", ', 0.9)')
        
        color = _hex_to_rgb(color)
        
        fig.add_trace(go.Scatter(
            x=n_bands, y=y_mean + y_std,
            mode='lines',
            line=dict(width=0),
            showlegend=False,
            hoverinfo='skip'
        ))

        fig.add_trace(go.Scatter(
            x=n_bands, y=y_mean - y_std,
            fill='tonexty',  # preenche entre esse trace e o anterior
            fillcolor=color.replace("0.9", '0.3'),
            line=dict(width=0),
            mode='lines',
            name=f"{type} ±1σ",
            hoverinfo='skip'
        ))

        # Curva da média
        fig.add_trace(go.Scatter(
            x=n_bands, y=y_mean,
            mode='lines',
            name=f"{type} (mean)",
            line=dict(color=color, width=2),
        ))

    # Layout
    fig.update_layout(
        title=description,
        xaxis_title="Índice de Banda",
        yaxis_title="Intensidade",
        template="plotly_white"
    )

    return fig
```

```python
def plot_sankey(df):
    
    cols = ['specie_type', 'region', 'beverage_type', 'beverage_subtype']
    all_labels = []
    for col in cols:
        all_labels.extend(df[col].unique())
    all_labels = list(pd.unique(all_labels))  # garantir que são únicos

    label_to_index = {label: idx for idx, label in enumerate(all_labels)}
    
    source = []
    target = []
    value = []
    
    for i in range(len(cols) - 1):
        # lógica de contagem 
        group = df.groupby([cols[i], cols[i+1]]).size().reset_index(name='count')

        for _, row in group.iterrows():
            source.append(label_to_index[row[cols[i]]])
            target.append(label_to_index[row[cols[i+1]]])
            value.append(row['count'])
            
    node_in_values = np.zeros(len(all_labels), dtype=int)
    for tgt, val in zip(target, value):
        node_in_values[tgt] += val
        
    node_hovertext = [
        f"{label}<br>Número de Amostras: {val} amostras" if val > 0 else label
        for label, val in zip(all_labels, node_in_values)
    ]
            
    fig = go.Figure(data=[go.Sankey(
        arrangement='snap',
        valueformat=",.0f",
        valuesuffix=" amostras",
        
        node=dict(
            pad=35,
            thickness=30,
            line=dict(color="#2b2d42", width=0.8),
            label=all_labels,
            color="#2b2d42",
            customdata=node_hovertext,       
            hovertemplate="%{customdata}<extra></extra>"
        ),
        
        link=dict(
            source=source,
            target=target,
            value=value,
            hovertemplate=(
                        "%{source.label} – %{target.label}<br>"   
                        "Número de amostras: %{value}<extra></extra>" 
            )
        )
    )])

    fig.update_layout(
        title_text="Fluxo de especiação, bebida, qualidade e região",
        font_size=10,
        template="plotly_white",
        width=1400,    
        height=650,
        margin=dict(l=10, r=10, b=10),
        
    )

    return fig
```

```python
def plot_sample_variability(
    df: pd.DataFrame,
    sample_size: int = 25,
    random_state: Optional[int] = None,
    color_by: Optional[str] = None,
    **filters
):
    """
    Plota espectros individuais de uma subamostra, sem média nem desvio padrão.

    Parâmetros
    ----------
    df : DataFrame
        Dados contendo a coluna 'spectrum' (lista/array) e colunas de metadados.
    sample_size : int, default 25
        Quantos espectros sortear para plotar. Usa todos se o grupo for menor.
    random_state : int ou None
        Semente do gerador de números aleatórios (reprodutibilidade).
    **filters : coluna=valor
        Filtros dinâmicos, como beverage_type='especial', region='mogiana' etc.
    """
    # 1 ─── Filtragem dinâmica
    sub = df.copy()
    for col, val in filters.items():
        if isinstance(val, list):
            sub = sub[sub[col].isin(val)]
        else:
            sub = sub[sub[col] == val]

    total = len(sub)
    if total == 0:
        raise ValueError("Nenhum registro corresponde aos filtros fornecidos.")

    # 2 ─── Sorteia amostra
    if total > sample_size:
        sub = sub.sample(n=sample_size, random_state=random_state)

    # 3 ─── Empilha espectros
    subsample = sub["spectrum"].values
    labels = sub['beverage_subtype'].values
    x = np.arange(len(subsample[0]))  # Eixo x baseado no comprimento de um espectro

    if color_by:
        labels = sub[color_by].values
        unique_labels = pd.unique(labels)
        color_palette = px.colors.qualitative.Safe 
        color_map = {label: color_palette[i % len(color_palette)] for i, label in enumerate(unique_labels)}
    else:
        labels = np.array([""] * len(subsample))
        color_map = {}
    

    # 4 ─── Plota cada espectro individualmente
    fig = go.Figure()
    
    for i in range(len(subsample)):
        label = labels[i]
        color = color_map[label] if color_by else "#1d2cf0"
        
        fig.add_trace(go.Scatter(
            x=x,
            y=subsample[i],
            mode="lines",
            line=dict(width=1, color=color),
            showlegend=(sample_size <= 15), # Mostra legenda apenas se poucos espectros
            hovertemplate=(
                f"{color_by}: {label}<br>"
            ) if color_by else (
                ''
            )
        ))

    # Layout
    title_filters = ", ".join(f"{k}={v}" for k, v in filters.items()) or "todos"
    fig.update_layout(
        title=f"Espectros Individuais ({title_filters}) — {len(sub)}/{total} exibidos",
        xaxis_title="Índice de Banda",
        yaxis_title="Intensidade",
        template="plotly_white",
        showlegend=(sample_size <= 15)
    )

    return fig, total
```

```python
def plot_sample_variability(
    df: pd.DataFrame,
    sample_size: int = 25,
    random_state: Optional[int] = None,
    color_by: Optional[str] = None,
    **filters
):
    """
    Plota espectros individuais de uma subamostra, sem média nem desvio padrão.

    Parâmetros
    ----------
    df : DataFrame
        Dados contendo a coluna 'spectrum' (lista/array) e colunas de metadados.
    sample_size : int, default 25
        Quantos espectros sortear para plotar. Usa todos se o grupo for menor.
    random_state : int ou None
        Semente do gerador de números aleatórios (reprodutibilidade).
    **filters : coluna=valor
        Filtros dinâmicos, como beverage_type='especial', region='mogiana' etc.
    """
    import numpy as np
    import pandas as pd
    import plotly.graph_objs as go
    import plotly.express as px

    # 1 ─── Filtragem dinâmica
    sub = df.copy()
    for col, val in filters.items():
        if isinstance(val, list):
            sub = sub[sub[col].isin(val)]
        else:
            sub = sub[sub[col] == val]

    total = len(sub)
    if total == 0:
        raise ValueError("Nenhum registro corresponde aos filtros fornecidos.")

    # 2 ─── Sorteia amostra
    if total > sample_size:
        sub = sub.sample(n=sample_size, random_state=random_state)

    # 3 ─── Empilha espectros
    subsample = sub["spectrum"].values
    labels = sub[color_by].values if color_by else np.array([""] * len(sub))
    x = np.arange(len(subsample[0]))  # Eixo x baseado no comprimento de um espectro

    if color_by:
        unique_labels = pd.unique(labels)
        color_palette = px.colors.qualitative.Safe
        color_map = {label: color_palette[i % len(color_palette)] for i, label in enumerate(unique_labels)}
    else:
        color_map = {}

    # 4 ─── Plota cada espectro individualmente
    fig = go.Figure()

    shown_legend = set()  # Para controlar quais rótulos já apareceram na legenda

    for i in range(len(subsample)):
        label = labels[i]
        color = color_map[label] if color_by else "#1d2cf0"
        show_legend = label not in shown_legend if color_by else False

        fig.add_trace(go.Scatter(
            x=x,
            y=subsample[i],
            mode="lines",
            line=dict(width=1, color=color),
            name=label if show_legend else None,
            showlegend=show_legend,
            hovertemplate=f"{color_by}: {label}<br>" if color_by else ''
        ))

        if show_legend:
            shown_legend.add(label)

    # Layout
    title_filters = ", ".join(f"{k}={v}" for k, v in filters.items()) or "todos"
    fig.update_layout(
        title=f"Espectros Individuais ({title_filters}) — {len(sub)}/{total} exibidos",
        xaxis_title="Índice de Banda",
        yaxis_title="Intensidade",
        template="plotly_white",
        showlegend=True
    )

    return fig, total
```

### X-axis Peak Drift Analysis for Beverage Type

```python
### Code for finding peaks

peak_distance = 50
peak_prominence = 0.08
peak_height = 0.15

def find_most_significant_peaks(spectrum: pd.Series,
                                peaks: np.ndarray) -> list:
    
    peak_intensities = spectrum.iloc[peaks]
    sorted_peaks = peak_intensities.sort_values(ascending=False).index[:2]
    top_2_peaks = list(sorted_peaks)  
    return top_2_peaks

def first_n_second_peak(row):
    
    spectrum = pd.Series(row['spectrum'])
    
    peaks, _ = find_peaks(spectrum,
                        distance=peak_distance, 
                        prominence=peak_prominence, 
                        height=peak_height)
    
    local_peaks_idx = find_most_significant_peaks(spectrum, peaks)
    local_peaks = []
    for peak_idx in local_peaks_idx:
        local_peaks.append(pd.Series(WAVELENGHTS).iloc[peak_idx] if peak_idx > -1 else -1)
        
    local_peaks.sort()
    if len(local_peaks) < 2:
        print(row.values)
        return
    p1_position = local_peaks[0]
    p2_position = local_peaks[1]
    p1_intensity = spectrum[local_peaks_idx[0]]
    p2_intensity = spectrum[local_peaks_idx[1]]
    
    return pd.Series([p1_position, p2_position, p1_intensity, p2_intensity],index=['peak_1_position', 'peak_2_position', 'peak_1_intensity', 'peak_2_intensity'])

def groupby_violin(variavel_interesse: str,
                   df: pd.Series) -> px.violin:
    
    color_sequence = list(COLORS['beverage_type'].values())
    
    fig = px.violin(
        df,
        x='beverage_type',
        y=f'{variavel_interesse}',
        box=True,
        color='beverage_type',
        color_discrete_sequence=color_sequence,
        labels={"beverage_type": "Tipos de bebidas"}
    )

    fig.update_layout(
        xaxis_title="Beverage Types",
        yaxis_title=f"{variavel_interesse}",
        # paper_bgcolor= 'rgba(0,0,0,0)',
        # font=dict(color='black')
        template='plotly_white'
        
    )
    
    return fig
```

### First Peak Area analysis

```python
WAVELENGHTS = np.asarray(WAVELENGHTS)
INIT_IDX = 60
INIT_WV = WAVELENGHTS[60]

def wl2idx(target_wl: float) -> int:
    return int(np.abs(WAVELENGHTS - target_wl).argmin())

def first_peak_area(row):
    peak_wl = row['peak_1_position']
    delta = peak_wl - INIT_WV
    end_wv = peak_wl + delta
    
    end_idx = wl2idx(end_wv)
    spec = np.asarray(row['spectrum'])          

    # integração trapezoidal área do pico
    area = trapezoid(spec[INIT_IDX:end_idx+1],
                     x=WAVELENGHTS[INIT_IDX:end_idx+1])

    return pd.Series(
        [INIT_IDX, end_idx, area],
        index=['peak1_start_idx', 'peak1_end_idx', 'peak1_area']
    )
```

```python
area = peaks.copy()
area[['peak1_start_idx', 'peak1_end_idx', 'peak1_area']] = peaks.apply(first_peak_area, axis=1)
```

```python
# feature functions
def find_most_significant_peaks(spectrum: pd.Series,
                                peaks: np.ndarray) -> list:
    """Seleciona os cinco picos mais significativos com base na intensidade.

    Args:
        spectrum (pd.Series): Espectro normalizado da amostra.
        peaks (np.ndarray): Array de índices dos picos detectados pelo método find_peaks().

    Returns:
        list: Lista de até cinco índices dos picos mais significativos, ordenados por intensidade.
              Se houver menos de cinco picos, os espaços restantes são preenchidos com -1.
    """
    peak_intensities = spectrum.iloc[peaks]
    sorted_peaks = peak_intensities.sort_values(ascending=False).index[:5]
    top_5_peaks = list(sorted_peaks) + [-1] * (5 - len(sorted_peaks))  
    return top_5_peaks

def normalize_sample(spectrum_sample):
    """Realiza a normalização Min-Max do espectro.
    Args:
        spectrum_sample (pd.Series): Espectro bruto da amostra.
    Returns:
        pd.Series: Espectro normalizado com valores entre 0 e 1.
    """
    spectrum_sample = spectrum_sample.astype(float)
    norm_spectrum = (spectrum_sample - spectrum_sample.min()) / (spectrum_sample.max() - spectrum_sample.min())
    return norm_spectrum

def iterate_over_peaks(wavelengths: pd.Series,
                       norm_spectrum: pd.Series,
                       local_peaks: list) -> dict:
    """Calcula a diferença posicional e a distância euclidiana entre os picos detectados.

    Args:
        wavelengths (pd.Series): Série contendo os comprimentos de onda correspondentes aos índices do espectro.
        norm_spectrum (pd.Series): Espectro normalizado da amostra.
        local_peaks (list): Lista contendo os índices dos picos detectados.

    Returns:
        dict: Dicionário contendo as diferenças posicionais no eixo do comprimento de onda
              e as distâncias euclidianas entre os pares de picos detectados.
    """
    distances = {}

    for i, j in combinations(range(5), 2):
        key_pos = f"pos_diff_{i+1}_{j+1}_peaks"
        key_euc = f"euclidean_dist_{i+1}_{j+1}_peaks"

        if local_peaks[i] != -1 and local_peaks[j] != -1:
            x1 = wavelengths.iloc[local_peaks[i]]
            x2 = wavelengths.iloc[local_peaks[j]]
            y1 = norm_spectrum.iloc[local_peaks[i]]
            y2 = norm_spectrum.iloc[local_peaks[j]]

            # Diferença posicional (somente eixo X)
            distances[key_pos] = abs(x1 - x2)

            # Distância euclidiana
            distances[key_euc] = np.sqrt((x2 - x1)**2 + (y2 - y1)**2)
        else:
            # Caso não existam 5 picos, padronizamos com -1
            distances[key_pos] = -1
            distances[key_euc] = -1

    return distances

def extract_spectral_features(wavelengths: pd.Series,
                              spectrum: pd.Series) -> dict:
    """Extrai diversas características do espectro, incluindo estatísticas globais,
       detecção de picos, larguras médias e distâncias entre picos.

    Args:
        wavelengths (pd.Series): Série contendo os comprimentos de onda.
        spectrum (pd.Series): Série contendo as intensidades espectrais.

    Returns:
        dict: Dicionário contendo todas as features extraídas do espectro.
    """
    
    features = {}
    norm_spectrum = normalize_sample(spectrum)    

    # Estatísticas básicas
    features["intensity_mean"] = norm_spectrum.mean()
    features["intensity_median"] = norm_spectrum.median()
    features["intensity_std"] = norm_spectrum.std()
    
    # Índice do pico de maior intensidade
    intensity_max_ind = norm_spectrum.idxmax()
    features["max_intensity_index"] = wavelengths.iloc[intensity_max_ind] if pd.notna(intensity_max_ind) else 0
    
    # Área sob a curva (integral)
    features["area_under_curve"] = trapezoid(norm_spectrum, wavelengths)

    # Encontra os picos locais
    
    # Retorna os índices dos picos, deve ser transformado em comprimento de onda
    peaks, _ = find_peaks(norm_spectrum,
                          distance=8,
                          prominence=0.1,
                          height=0.15)
    
    local_peaks = find_most_significant_peaks(norm_spectrum, peaks)
    for i, peak_idx in enumerate(local_peaks):
        features[f"peak_{i+1}_position"] = wavelengths.iloc[peak_idx] if peak_idx > -1 else -1
        features[f"peak_{i+1}_intensity"] = norm_spectrum.iloc[peak_idx] if peak_idx > -1 else -1

    # Largura média para os dois maiores picos
    valid_peaks = [p for p in local_peaks if p != -1]
    if len(valid_peaks) >= 2: 
        widths, width_heights, left_ips, right_ips = peak_widths(norm_spectrum, valid_peaks[:2], rel_height=0.5)
        for i in range(len(widths)):
            features[f"peak_{i+1}_width"] = (wavelengths.iloc[int(right_ips[i])] - wavelengths.iloc[int(left_ips[i])]) if i < len(right_ips) else -1
    else:
        features["peak_1_width"] = -1
        features["peak_2_width"] = -1

    features.update(iterate_over_peaks(wavelengths, norm_spectrum, local_peaks))
    
    return features   

def vis_sample_features(wavelengths: pd.Series,
                        spectrum: pd.Series,
                        sample_class: str,
                        features: dict):
    fig = go.Figure()
    norm_spectrum = normalize_sample(spectrum)  
    peaks, _ = find_peaks(norm_spectrum,
                          distance=8,
                          prominence=0.1,
                          height=0.15)
    
    local_peaks = find_most_significant_peaks(norm_spectrum, peaks)
    valid_peaks = [p for p in local_peaks if p != -1]
    
    #spectrum vis
    fig.add_trace(go.Scatter(
        x=wavelengths,
        y=norm_spectrum,
        mode='lines',
        name='Espectro',
        marker=dict(color='#1d2cf0'),
        opacity=0.7
    ))
    
    if valid_peaks: 
        fig.add_trace(go.Scatter(
            x=wavelengths.iloc[valid_peaks],  
            y=norm_spectrum.iloc[valid_peaks],
            mode='markers',
            marker=dict(color='#03071e', size=10),
            name=f'Picos da amostra'
        ))
        
        widths, width_heights, left_ips, right_ips = peak_widths(norm_spectrum, valid_peaks[:2], rel_height=0.5)
        left_x = np.interp(left_ips, np.arange(len(wavelengths)), wavelengths)
        right_x = np.interp(right_ips, np.arange(len(wavelengths)), wavelengths)
        
        for i in range(len(widths)):  
            fig.add_trace(go.Scatter(
            x=[left_x[i], right_x[i]],  
            y=[width_heights[i], width_heights[i]],  
            mode='lines',
            line=dict(color='#03071e'),
        name=f'Meia Largura Pico {i+1}'
            ))
    
    for i in range(1, 6):
        for j in range(i + 1, 6):
            pos_diff_key = f"pos_diff_{i}_{j}_peaks"
            euclidean_key = f"euclidean_dist_{i}_{j}_peaks"
            
            if pos_diff_key in features and features[pos_diff_key] > -1:
                peak_1 = features[f"peak_{i}_position"]
                peak_2 = features[f"peak_{j}_position"]
            
                if peak_1 > -1 and peak_2 > -1:
                    fig.add_trace(go.Scatter(
                        x=[peak_1, peak_2],
                        y=[features[f"peak_{i}_intensity"], features[f"peak_{j}_intensity"]],
                        mode='lines',
                        line=dict(color='#03071e', dash='dot'),
                        name=f'Distância Euclidiana {i}-{j}'
                    ))
    
    fig.update_layout(
        title=f'Sample Features Visualization{sample_class}',
        xaxis_title="Comprimento de Onda (nm)",
        yaxis_title="Intensidade",
        template="plotly_white",
    )
    return fig
```
