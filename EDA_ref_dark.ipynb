```python
WAVELENGHTS = [317.998, 320.704, 323.408, 326.11 , 328.809, 331.505, 334.198,
       336.889, 339.577, 342.262, 344.944, 347.623, 350.3  , 352.973,
       355.643, 358.31 , 360.974, 363.635, 366.293, 368.947, 371.598,
       374.246, 376.89 , 379.531, 382.168, 384.802, 387.433, 390.06 ,
       392.683, 395.302, 397.918, 400.531, 403.139, 405.743, 408.344,
       410.941, 413.534, 416.123, 418.708, 421.289, 423.866, 426.439,
       429.008, 431.572, 434.133, 436.689, 439.241, 441.788, 444.332,
       446.87 , 449.405, 451.935, 454.461, 456.982, 459.498, 462.01 ,
       464.518, 467.021, 469.519, 472.012, 474.501, 476.985, 479.464,
       481.938, 484.408, 486.873, 489.332, 491.787, 494.237, 496.682,
       499.122, 501.557, 503.987, 506.412, 508.831, 511.246, 513.655,
       516.059, 518.458, 520.852, 523.24 , 525.623, 528.001, 530.373,
       532.74 , 535.102, 537.458, 539.809, 542.154, 544.494, 546.828,
       549.157, 551.48 , 553.797, 556.109, 558.416, 560.716, 563.011,
       565.301, 567.584, 569.862, 572.135, 574.401, 576.662, 578.916,
       581.165, 583.409, 585.646, 587.877, 590.103, 592.322, 594.536,
       596.744, 598.946, 601.141, 603.331, 605.515, 607.693, 609.865,
       612.03 , 614.19 , 616.343, 618.491, 620.632, 622.767, 624.896,
       627.019, 629.136, 631.247, 633.351, 635.449, 637.541, 639.627,
       641.707, 643.78 , 645.847, 647.908, 649.963, 652.011, 654.053,
       656.089, 658.118, 660.141, 662.158, 664.168, 666.172, 668.17 ,
       670.162, 672.147, 674.126, 676.098, 678.064, 680.024, 681.977,
       683.924, 685.864, 687.798, 689.726, 691.648, 693.563, 695.471,
       697.373, 699.269, 701.158, 703.041, 704.918, 706.788, 708.652,
       710.509, 712.36 , 714.205, 716.043, 717.874, 719.7  , 721.519,
       723.331, 725.137, 726.937, 728.731, 730.518, 732.298, 734.073,
       735.84 , 737.602, 739.357, 741.106, 742.848, 744.585, 746.314,
       748.038, 749.755, 751.466, 753.17 , 754.869, 756.561, 758.246,
       759.926, 761.599, 763.266, 764.927, 766.581, 768.229, 769.871,
       771.507, 773.137, 774.76 , 776.378, 777.989, 779.594, 781.193,
       782.786, 784.372, 785.953, 787.528, 789.096, 790.659, 792.215,
       793.766, 795.31 , 796.849, 798.382, 799.908, 801.429, 802.944,
       804.453, 805.956, 807.454, 808.945, 810.431, 811.911, 813.385,
       814.854, 816.316, 817.774, 819.225, 820.671, 822.111, 823.546,
       824.975, 826.398, 827.816, 829.229, 830.636, 832.037, 833.433,
       834.824, 836.209, 837.589, 838.964, 840.333, 841.697, 843.056,
       844.41 , 845.759, 847.102, 848.44 , 849.773, 851.102, 852.425,
       853.743, 855.056, 856.364, 857.668, 858.966, 860.26 , 861.548,
       862.833, 864.112, 865.387, 866.657, 867.922, 869.183, 870.439,
       871.691, 872.938, 874.181, 875.419, 876.653, 877.883, 879.108,
       880.33 , 881.547, 882.759, 883.968, 885.173, 886.373, 887.57 ,
       888.762]

def extract_metadata(txt_path: str) -> dict:
    
    metadata = {"timestamp": None}
    
    if not os.path.exists(txt_path):
        return metadata
    
    with open(txt_path, "r", encoding="utf-8") as file:
        for line in file:
            line = line.strip()
            if line.startswith("Date ="):
                date = line.split("=", 1)[1].strip()
                timestamp = datetime.datetime.strptime(date, "%Y/%m/%d %H:%M:%S")
                metadata['timestamp'] = timestamp.isoformat(timespec="seconds")
    return metadata

NOVAS_MEDIDAS_PATH = "data\Medidas de TODOS os grãos disponíveis\Medidas de TODOS os grãos disponíveis"

regions_ref = ['mogiana', 'minas gerais', 'florão', 'sao paulo', 'rondônia']
fine_class_ref = ['mole', 'dura', '84-85', 'moka', '7x8']

list_of_dictionaries = []

for dir in os.listdir(NOVAS_MEDIDAS_PATH):
    PATH_SPECIES = os.path.join(NOVAS_MEDIDAS_PATH, dir)

    # Espécie/ bebida dos arquivos do diretório
    split_dir = dir.split(' ')
    especie = split_dir[0]
    bebida = split_dir[1]
    
    if not os.path.isdir(PATH_SPECIES):
        continue
    
    for dir_regions in os.listdir(PATH_SPECIES):
        PATH_REGIONS = os.path.join(PATH_SPECIES, dir_regions)
        if not os.path.isdir(PATH_REGIONS):
            continue
        
        # Região dos .csv do diretório

        region = 'nao informado'
        for region_ref in regions_ref:
            if region_ref.lower() in dir_regions.lower():
                region = region_ref
                break
        
        files = os.listdir(PATH_REGIONS)
        
        # Garantir que nenhum dos arquivos _Dark ou _ref sejam guardados no df
        csv_files = [
            f for f in files
            if f.endswith(".csv") and "Dark" not in f and "ref" not in f]
        
        for csv_file in csv_files:
            sample = {}
            
            csv_path = os.path.join(PATH_REGIONS, csv_file)
            base_file_name = os.path.splitext(csv_file)[0]
            txt_file = base_file_name + ".txt"
            txt_path = os.path.join(PATH_REGIONS, txt_file)

            spectrum = pd.read_csv(csv_path, usecols=['Average'])['Average']

            # Procurar espectros dark e ref correspondentes
            dark_file = base_file_name + "_Dark.csv"
            ref_file = base_file_name + "_ref.csv"
            
            dark_path = os.path.join(PATH_REGIONS, dark_file)
            ref_path = os.path.join(PATH_REGIONS, ref_file)
            
            if os.path.exists(dark_path):
                dark_spectrum = pd.read_csv(dark_path, usecols=['Average'])['Average'].tolist()
            else:
                dark_spectrum = None

            if os.path.exists(ref_path):
                ref_spectrum = pd.read_csv(ref_path, usecols=['Average'])['Average'].tolist()
            else:
                ref_spectrum = None

            metadata = extract_metadata(txt_path)

            sample_fine_class = ''
            for fine_class in fine_class_ref:
                if fine_class in csv_file.lower():
                    sample_fine_class = fine_class
                    break
            else:
                sample_fine_class = bebida

            sample['timestamp'] = metadata['timestamp']
            sample['region'] = region.lower()
            sample['spectrum'] = spectrum.tolist()
            sample['dark_spectrum'] = dark_spectrum
            sample['ref_spectrum'] = ref_spectrum
            sample['specie_type'] = especie.lower()
            sample['composition'] = (
                '50/50' if '50x50' in csv_file
                else '70/30' if '+' in csv_file
                else 'puro'
            )
            sample['beverage_type'] = (
                'blend' if bebida == '-'
                else bebida.lower()
            )
            sample['fine class'] = (
                'blend' if sample_fine_class == '-'
                else sample_fine_class.lower()
            )

            list_of_dictionaries.append(sample)
            


df = pd.DataFrame(list_of_dictionaries)

```
